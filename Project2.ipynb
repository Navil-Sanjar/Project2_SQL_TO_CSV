```python
import psycopg2
import pandas as pd
```

```python
# Database connection config

conn=psycopg2.connect(host="localhost",database="demodb",user="postgres",password=1234)
```

```python
# Sample SQL queries

queries = {"Top 3 Employee":""" SELECT employee_name,salary from employees ORDER BY salary DESC LIMIT 3""",
           "Average Salary per Department":""" SELECT department,AVG(salary) as Average_Salary from employees GROUP BY department """,
        "Most Populated Department":"""SELECT department,count(*) as Employee_Count from employees GROUP BY department ORDER BY Employee_Count DESC LIMIT 1"""}
```

```python
# Run and export all queries
with conn:
    for title,sql in queries.items():
        df=pd.read_sql(sql,conn)
        print(f"\n 🔹{title}\n",df)
        df.to_csv(f"../reports/{title.replace(' ','_').lower()}.csv",index=False)
        
conn.close()
```
